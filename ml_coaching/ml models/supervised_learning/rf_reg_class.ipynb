```python
import numpy as np 
import pandas as pd 
```

```python
df = pd.read_csv(r"D:\py_machine_learning\datasets\insurance - insurance.csv")
df
```

```python
x = df.drop(columns=['charges'])
y = df['charges']
```

```python
from sklearn.preprocessing import LabelEncoder
```

```python
lb = LabelEncoder()
```

```python
x['sex'] = lb.fit_transform(df['sex'])
x['smoker'] = lb.fit_transform(df['smoker'])
x['region'] = lb.fit_transform(df['region'])
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
from sklearn.ensemble import RandomForestRegressor
```

```python
dt = RandomForestRegressor()
```

```python
dt.fit(x_train,y_train)
```

```python
y_pred = dt.predict(x_test)
```

```python
from sklearn.metrics import r2_score
```

```python
r2_score(y_test,y_pred)
```

```python
cf = pd.read_csv(r"D:\py_machine_learning\datasets\covid_toy - covid_toy.csv")
cf
```

```python
cf['fever'] = cf['fever'].fillna(cf['fever'].mean())
```

```python
lb = LabelEncoder()
```

```python
cf['gender'] = lb.fit_transform(cf['gender'])
cf['cough'] = lb.fit_transform(cf['cough'])
cf['city'] = lb.fit_transform(cf['city'])
```

```python
x = cf.drop(columns=['has_covid'])
y = cf['has_covid']
```

```python
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
from sklearn.ensemble import RandomForestClassifier
```

```python
rf = RandomForestClassifier()
```

```python
rf.fit(x_train,y_train)
```

```python
y_pred = rf.predict(x_test)
```

```python
from sklearn.metrics import accuracy_score
```

```python
accuracy_score(y_test,y_pred)
```
