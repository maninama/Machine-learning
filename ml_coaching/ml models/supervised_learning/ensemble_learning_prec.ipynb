```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
from sklearn.ensemble import RandomForestClassifier
```

```python
df = pd.read_csv(r"D:\py_machine_learning\datasets\covid_toy - covid_toy.csv")
df.head(2)
```

```python
df['fever'] = df['fever'].fillna(df['fever'].mean())
```

```python
df.isnull().sum()
```

```python
from sklearn.preprocessing import LabelEncoder
```

```python
lb = LabelEncoder()
```

```python
df['gender'] = lb.fit_transform(df['gender'])
df['cough'] = lb.fit_transform(df['cough'])
df['city'] = lb.fit_transform(df['city'])
```

```python
x = df.drop(columns=['has_covid'])
y = df['has_covid']
```

```python
x_train, x_test, y_train, y_test = train_test_split(x,y,test_size=0.20,random_state =42)
```

```python
base_learners = []

num_base_learners = 5

for _ in range(num_base_learners):
  
  bootstrap_indices = np.random.choice(len(x_train), size=len(x_train), replace=True)

  X_bootstrap = x_train.iloc[bootstrap_indices]
  y_bootstrap = y_train.iloc[bootstrap_indices]

 
  base_learner = RandomForestClassifier(n_estimators=10, random_state=42)
  base_learner.fit(X_bootstrap, y_bootstrap)

  base_learners.append(base_learner)

base_predictions = []
for base_learner in base_learners:
  y_pred = base_learner.predict(x_test)
  base_predictions.append(y_pred)


ensemble_predictions = np.round(np.mean(base_predictions, axis=0))


accuracy = accuracy_score(y_test, ensemble_predictions)
print("Ensemble Accuracy:", accuracy)
```
