```python
import pandas as pd 
from sklearn.preprocessing import LabelEncoder, OrdinalEncoder, OneHotEncoder,StandardScaler
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression 
```

```python
df = pd.read_csv(r"D:\py_machine_learning\datasets\covid_toy - covid_toy.csv")

print(df.head())
```

```python
df["fever"] = df["fever"].fillna(df["fever"].mean())

x = df.drop("has_covid", axis=1)
y = df["has_covid"]
```

```python
le = LabelEncoder()

x["gender"] = le.fit_transform(x["gender"])
x["city"] = le.fit_transform(x["city"])

x.head()
```

```python
oe = OrdinalEncoder(categories=[["Mild", "Strong"]])

x["cough"] = oe.fit_transform(x[["cough"]])

x.head()
```

```python
ohe = OneHotEncoder(sparse_output=False)

ohe_encode = ohe.fit_transform(x[["gender", "city"]])

ohe_encode
```

```python
dummy_df = pd.get_dummies(
    x,
    columns=["gender", "city"]
)
dummy_df.head()
```

```python
numeric_cols = ["age", "fever"]
ordinal_cols = ["cough"]
nominal_cols = ["gender", "city"]
```

```python

from sklearn.compose import ColumnTransformer

preprocessor = ColumnTransformer(
    transformers=[
        ("num", StandardScaler(), numeric_cols),
        ("ord", OrdinalEncoder(categories=oe), ordinal_cols),
        ("nom", OneHotEncoder(drop="first"), nominal_cols)
    ]
)

model = Pipeline([
    ("preprocessing", preprocessor),
    ("classifier", LogisticRegression())
])

model
```
